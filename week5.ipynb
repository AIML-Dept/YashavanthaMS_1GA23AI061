```python
# ============================================================
# EASY: Hadamard Superposition - Shot Convergence
# ============================================================

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

simulator = AerSimulator()

# Create |+> state
qc = QuantumCircuit(1, 1)
qc.h(0)
qc.measure(0, 0)

print("EASY: Hadamard Superposition Circuit")
print("-------------------------------------")
print(qc.draw())

print("\nMeasurement Results")
print("-------------------")
print("Shots\tP(0)\tP(1)")

for shots in [100, 1000, 10000]:

    result = simulator.run(qc, shots=shots).result()
    counts = result.get_counts()

    p0 = counts.get('0', 0) / shots
    p1 = counts.get('1', 0) / shots

    print(f"{shots}\t{p0:.4f}\t{p1:.4f}")
```

```python
# ============================================================
# MEDIUM: X-Basis Measurement of |+> and |->
# ============================================================

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

simulator = AerSimulator()
shots = 1000

# -------------------- |+> --------------------
plus = QuantumCircuit(1, 1)

# Prepare |+>
plus.h(0)

# Transform X-basis to computational basis
plus.h(0)

# Measure
plus.measure(0, 0)

print("MEDIUM: |+> X-Basis Measurement")
print("--------------------------------")
print(plus.draw())

result = simulator.run(plus, shots=shots).result()
print("Result:", result.get_counts())


# -------------------- |-> --------------------
minus = QuantumCircuit(1, 1)

# Prepare |->
minus.x(0)
minus.h(0)

# Transform X-basis to computational basis
minus.h(0)

# Measure
minus.measure(0, 0)

print("\nMEDIUM: |-> X-Basis Measurement")
print("--------------------------------")
print(minus.draw())

result = simulator.run(minus, shots=shots).result()
print("Result:", result.get_counts())

print("\nInterpretation:")
print("|+> gives approximately 0 with probability 1.")
print("|-> gives approximately 1 with probability 1.")
```

```python
# ============================================================
# HARD: Partial Measurement of a Bell Pair
# ============================================================

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

simulator = AerSimulator()

# Bell pair
qc = QuantumCircuit(2, 1)

# Create Bell state
qc.h(0)
qc.cx(0, 1)

# Measure ONLY qubit 0
qc.measure(0, 0)

print("HARD: Bell Pair Partial Measurement")
print("------------------------------------")
print(qc.draw())

result = simulator.run(qc, shots=1000).result()
counts = result.get_counts()

print("\nMeasurement Results:")
print(counts)

print("\nAnalysis:")
print("If q0 = 0, the unmeasured q1 collapses to |0>.")
print("If q0 = 1, the unmeasured q1 collapses to |1>.")
print("Therefore, the two qubits remain perfectly correlated.")
```

```python
# ============================================================
# REAL-WORLD: Estimate Unknown Rotation Angle
# Simple Quantum State Tomography
# ============================================================

import numpy as np
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

simulator = AerSimulator()

# Unknown rotation angle
true_theta = np.pi / 3

shots = 5000

qc = QuantumCircuit(1, 1)

# Apply unknown Ry(theta) rotation
qc.ry(true_theta, 0)

# Measure
qc.measure(0, 0)

print("REAL-WORLD: Unknown Rotation")
print("-----------------------------")
print(qc.draw())

result = simulator.run(qc, shots=shots).result()
counts = result.get_counts()

# Probability of measuring |1>
p1 = counts.get('1', 0) / shots

# P(1) = sin²(theta/2)
estimated_theta = 2 * np.arcsin(np.sqrt(p1))

print("\nMeasurement Results:")
print(counts)

print("\nTomography Results:")
print(f"True angle      = {true_theta:.4f} radians")
print(f"Measured P(1)   = {p1:.4f}")
print(f"Estimated angle = {estimated_theta:.4f} radians")
```

```python
# ============================================================
# CHALLENGE: Quantum Randomness vs Biased Classical Generator
# Statistical Hypothesis Testing
# ============================================================

import numpy as np
from scipy.stats import chisquare
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

simulator = AerSimulator()

shots = 10000

# ------------------------------------------------------------
# Quantum Random Bit Generator
# ------------------------------------------------------------

quantum = QuantumCircuit(1, 1)

quantum.h(0)
quantum.measure(0, 0)

print("CHALLENGE: Quantum Random Bit Generator")
print("----------------------------------------")
print(quantum.draw())

result = simulator.run(quantum, shots=shots).result()
quantum_counts = result.get_counts()

q0 = quantum_counts.get('0', 0)
q1 = quantum_counts.get('1', 0)

print("\nQuantum Results:")
print(f"0 = {q0}")
print(f"1 = {q1}")


# ------------------------------------------------------------
# Biased Classical Generator
# ------------------------------------------------------------

classical_bits = np.random.choice(
    [0, 1],
    size=shots,
    p=[0.70, 0.30]
)

c0 = np.sum(classical_bits == 0)
c1 = np.sum(classical_bits == 1)

print("\nBiased Classical Results:")
print(f"0 = {c0}")
print(f"1 = {c1}")


# ------------------------------------------------------------
# Chi-Square Test
# H0: Probability of 0 and 1 is 50/50
# ------------------------------------------------------------

expected = [shots / 2, shots / 2]

quantum_test = chisquare(
    [q0, q1],
    f_exp=expected
)

classical_test = chisquare(
    [c0, c1],
    f_exp=expected
)

print("\nStatistical Hypothesis Test")
print("---------------------------")

print("\nQuantum Generator:")
print(f"Chi-square = {quantum_test.statistic:.4f}")
print(f"p-value    = {quantum_test.pvalue:.6f}")

print("\nClassical Generator:")
print(f"Chi-square = {classical_test.statistic:.4f}")
print(f"p-value    = {classical_test.pvalue:.6f}")


# ------------------------------------------------------------
# Conclusion
# ------------------------------------------------------------

alpha = 0.05

print("\nConclusion:")

if quantum_test.pvalue >= alpha:
    print("Quantum data is consistent with a 50/50 distribution.")
else:
    print("Quantum data shows significant bias.")

if classical_test.pvalue < alpha:
    print("Classical generator is statistically biased.")
else:
    print("Classical generator is consistent with 50/50.")
```

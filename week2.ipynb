```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# Create one qubit
qc = QuantumCircuit(1)

# Apply X Gate
qc.x(0)

# Statevector
state = Statevector.from_instruction(qc)

print("Quantum Circuit:")
print(qc)

print("\nStatevector:")
print(state)
```

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
import matplotlib.pyplot as plt

qc = QuantumCircuit(1)

# Hadamard Gate
qc.h(0)

# Phase Gate
qc.s(0)

print(qc)

state = Statevector.from_instruction(qc)

print("\nStatevector:")
print(state)

# Bloch Sphere
plot_bloch_multivector(state)
plt.show()
```

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
import numpy as np

qc = QuantumCircuit(2)

# Hadamard on both qubits
qc.h(0)
qc.h(1)

print(qc)

state = Statevector.from_instruction(qc)

print("\nStatevector:")
print(state)

# Probabilities
probabilities = np.abs(state.data)**2

print("\nProbabilities:")
print(probabilities)

print("\nSum of Probabilities:")
print(np.sum(probabilities))
```

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
import numpy as np
import matplotlib.pyplot as plt

# Random phase
phase = np.random.uniform(0, 0.5)

qc = QuantumCircuit(1)

qc.h(0)

# Phase error
qc.p(phase, 0)

print("Random Phase Error =", phase)

print(qc)

state = Statevector.from_instruction(qc)

print("\nStatevector:")
print(state)

plot_bloch_multivector(state)
plt.show()
```

```python
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
import numpy as np
import matplotlib.pyplot as plt

def create_qubit(theta, phi):

    alpha = np.cos(theta/2)

    beta = np.exp(1j*phi) * np.sin(theta/2)

    state = Statevector([alpha, beta])

    print("Statevector:")
    print(state)

    probabilities = np.abs(state.data)**2

    print("\nProbability of |0> =", probabilities[0])
    print("Probability of |1> =", probabilities[1])

    print("\nNormalization =", np.sum(probabilities))

    plot_bloch_multivector(state)
    plt.show()

# Example
theta = np.pi/3
phi = np.pi/4

create_qubit(theta, phi)
```
